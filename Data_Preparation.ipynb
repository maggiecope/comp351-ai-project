```python
import pandas as pd

weather = pd.read_csv("weather.csv")
waits   = pd.read_csv("disney_wait_times.csv")
```

```python
weather.head()
waits.head()
```

```python
# 1. Parse Local Time as timezone-aware datetimes in UTC
waits["Local Time"] = pd.to_datetime(
    waits["Local Time"],
    utc=True,
    errors="coerce"   # turns bad/missing values into NaT instead of breaking
)

# 2. Now it's a proper datetime64[ns, UTC] Series, so .dt works
waits["Date"] = waits["Local Time"].dt.strftime("%Y-%m-%d")
```

```python
weather_cols_to_keep = [
    "Date",
    "tempmax", "tempmin", "temp",
    "humidity", "precip",
    "windgust", "windspeed", "winddir",
    "sealevelpressure",
    "cloudcover", "visibility",
    "solarradiation", "solarenergy",
    "uvindex", "severerisk",
    "sunrise", "sunset", "moonphase",
    "conditions", "description",
]

weather_for_merge = weather[weather_cols_to_keep]
weather_for_merge.head()
```

```python
merged = waits.merge(
    weather_for_merge,
    on="Date",
    how="left"    # keep all wait-time rows even if some dates lack weather
)
```

```python
final_cols = [
    "Land",
    "Ride",
    "Wait Time",
    "Local Time",
    "Day of Week",
    "Date",
    "tempmax",
    "tempmin",
    "temp",
    "humidity",
    "precip",
    "windgust",
    "windspeed",
    "winddir",
    "sealevelpressure",
    "cloudcover",
    "visibility",
    "solarradiation",
    "solarenergy",
    "uvindex",
    "severerisk",
    "sunrise",
    "sunset",
    "moonphase",
    "conditions",
    "description",
]

merged = merged[final_cols]
merged.head()
```

```python
# Make sure Wait Time is numeric (just in case it's a string)
merged["Wait Time"] = pd.to_numeric(merged["Wait Time"], errors="coerce")

# Keep only rows where Wait Time is not 0 and not NaN
merged = merged[merged["Wait Time"] != 0].dropna(subset=["Wait Time"])

merged.head()
```

```python
merged.to_csv("wait_times_with_weather_clean.csv", index=False)
```
